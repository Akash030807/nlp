```python
import numpy as np

def manual_softmax(x):
    e_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e_x / np.sum(e_x, axis=-1, keepdims=True)

def manual_layer_norm(x, gamma, beta, epsilon=1e-6):
    mean = np.mean(x, axis=-1, keepdims=True)
    variance = np.var(x, axis=-1, keepdims=True)
    x_normalized = (x - mean) / np.sqrt(variance + epsilon)
    return gamma * x_normalized + beta
```

```python
d_model = 64
max_seq_len = 10
batch_size = 2

input_embeddings = np.random.rand(batch_size, max_seq_len, d_model)

position = np.arange(max_seq_len)[:, np.newaxis]
div_term = np.exp(np.arange(0, d_model, 2) * -(np.log(10000.0) / d_model))
pe = np.zeros((max_seq_len, d_model))
pe[:, 0::2] = np.sin(position * div_term);
pe[:, 1::2] = np.cos(position * div_term)
pe = pe[np.newaxis, :, :]

encoder_input = input_embeddings + pe
display(encoder_input)
```

```python
num_heads = 4
d_k = d_model // num_heads

W_q = np.random.rand(d_model, d_k * num_heads)
b_q = np.random.rand(d_k * num_heads)
W_k = np.random.rand(d_model, d_k * num_heads)
b_k = np.random.rand(d_k * num_heads)
W_v = np.random.rand(d_model, d_k * num_heads)
b_v = np.random.rand(d_k * num_heads)

Q = np.dot(encoder_input, W_q) + b_q
K = np.dot(encoder_input, W_k) + b_k
V = np.dot(encoder_input, W_v) + b_v

display(Q)
display(K)
display(V)
```

```python
heads_output = []
for i in range(num_heads):
    head_Q = Q[:, :, i * d_k : (i + 1) * d_k]
    head_K = K[:, :, i * d_k : (i + 1) * d_k]
    head_V = V[:, :, i * d_k : (i + 1) * d_k]

    scores = np.matmul(head_Q, head_K.transpose(0, 2, 1)) / np.sqrt(d_k)
    attention_weights = manual_softmax(scores)
    head_output = np.matmul(attention_weights, head_V)
    heads_output.append(head_output)

multi_head_output = np.concatenate(heads_output, axis=-1)

W_o = np.random.rand(d_model, d_model)
b_o = np.random.rand(d_model)
attention_output = np.dot(multi_head_output, W_o) + b_o

display(attention_output)
```

```python
gamma_ln1 = np.random.rand(d_model)
beta_ln1 = np.random.rand(d_model)

attention_add = encoder_input + attention_output

attention_norm_output = manual_layer_norm(attention_add, gamma_ln1, beta_ln1)

display(attention_norm_output)
```

```python
d_ff = 2048

W_ff1 = np.random.rand(d_model, d_ff)
b_ff1 = np.random.rand(d_ff)
W_ff2 = np.random.rand(d_ff, d_model)
b_ff2 = np.random.rand(d_model)

ffn_output_inner = np.maximum(0, np.dot(attention_norm_output, W_ff1) + b_ff1)

ffn_output = np.dot(ffn_output_inner, W_ff2) + b_ff2

display(ffn_output)
```

```python
gamma_ln2 = np.random.rand(d_model)
beta_ln2 = np.random.rand(d_model)

encoder_output_add = attention_norm_output + ffn_output

encoder_output = manual_layer_norm(encoder_output_add, gamma_ln2, beta_ln2)

display(encoder_output)
```
